## Methodes ndarray : MATHEMATIQUES

Ce sont les methodes applicables sur les tableaux a n dimensions qui sont donc des objets de cette classe 

```python
import numpy as np
```

```python
A = np.random.randint(0, 10, [2, 3]) 
A
```

```python
A.sum() #Somme de tous les elements du tableau A
```

```python
A.sum(axis=0) #Somme de tous les elements du tableau A selon l'axe 0 (les colonnes)
A.sum(axis=1) #Somme de tous les elements du tableau A selon l'axe 1 (les lignes)

## On peut egalement calculer la somme cumulee, le produit, le maximum, le minimum
## ... Et j'en passe 
A.prod()

## On peut aussi trouver la postion du maximum et du minimum dans le tableau A
A.argmax() #Position du maximum dans le tableau A
A.argmin(axis=0) #Position du minimum dans le tableau A

## On a aussi la methode argsort() qui permet de donner les indices  ou les positions des elements du tableau dans l'ordre croissant ou decroissant
A.argsort() #Indices des elements du tableau A dans l'ordre croissant
```

### NB : Il existe aussi d'autres methodes qui ne sont pas forcement specifiques **ndarray**  mais tout simplement de **Numpy**

## Les datasets ne sont pas toujours clean (rarement carrement), avec des valeurs qui non numeriques : nan values 

Bien qu'evidemment ce probleme est gerer dans Pandas, il existe quelques methodes Numpy qui permettent de faire des statisqtiques avec ces nan values sans complication 

Les methodes classiques sont 

```python
np.nanmean(A)
np.nanstd(A)
np.nanvar(A)
```

Comment compter de valeurs manquantes (nan values) dans un tableau numpy ?

```python
# On le fait avec la methode np.isnan() qui retourne un tableau de booleen et on applique la methode sum() pour avoir le nombre de nan values
np.isnan(A).sum()
```

Comment se debarasser des nan values dans son taleau numpy ? 

```python
#On va se servir de du boolean indexing 
A[np.isnan(A)] = 0 #On remplace les nan values par 0
```

## Exercice : STANDARDISATION D'UNE MATRICE 
### Soit la matrice A ci-dessous, standardisez la sur chaque colonne. Donc chaque colonne doit avoir moyenne = 0 et ecart type = 1 


```python
np.random.seed(0)
A = np.random.randint(0, 10, [10, 5])
A
```

## *Resolution de l'exercice*

Bon le principe est de calculer d'abord les diffenretes statistiques de chaque colonnes et ensuite determiner la standardisation

```python
A #matrice A (10, 5)
```

**Colonne 1**

```python
A[:, 0].mean() #Moyenne de la premiere colonne
A[:, 0].std() #Ecart type de la premiere colonne


(A - A[:, 0].mean())/A[:, 0].std()
```
