## Premier cours sur numpy 

Au lieu d'utiliser des listes classiques en python, on va plus plutot utiliser des 
***ndarray** qui sont tableaux (objet) en python de la biblio Numpy. On d'abord donc des tableaux a : 
- 1 dimension
- 2 dimensions
- 3 dimenions...
- n dimensions

Mais dans notre contexte, on travaillera le plus souvent avec des tableaux a deux dimensions. On peut donc y stocker des valeurs de meme type par ligne et par colonne. on peut egalement y stocker les pixels d'une image (valeur = 0-250). 

## Quelques attributs :

```python
ndim = 1
shape = (2, ) #C'est un tuple 
```

## Quelques contructeurs : 

```python
import numpy as np
```

```python
A = np.array([1, 2, 3]) ### Tab de dim 1 
A.ndim
A.shape
A.size #La taille du tableau, le nombre d'éléments dans le tableau

#########Initialisation d'un tableau avec des valeurs
np.full((2, 3), 5) #Initialisation d'un tableau de 2 lignes et 3 colonnes avec la valeur 5

np.zeros((2, 3)) #Initialisation d'un tableau de 2 lignes et 3 colonnes avec la valeur 0

np.ones((2, 3)) #Initialisation d'un tableau de 2 lignes et 3 colonnes avec la valeur 1

## Utilisation du module random pour générer des valeurs aléatoires
np.random.randn(2, 3) #Initialisation d'un tableau de 2 lignes et 3 colonnes avec des valeurs aléatoires suivant une loi normale

np.linspace(0, 1, 5) #Initialisation d'un tableau de 5 valeurs espacées uniformément entre 0 et 1
np.arange(0, 10, 1) #Initialisation d'un tableau de valeurs allant de 0 à 10 avec un pas de 1

### Specifier le type de données d'un tableau
np.array([1, 2, 3], dtype=np.float32) #Initialisation d'un tableau de valeurs entières avec un type de données float32
```

## Manipulation 
 On peut assembler les tableaux de deux manieres : 
 - **Horizontalement**
 - **Verticalement**

```python
B = np.array([4, 5, 6]) ### Tab de dim 1
# On utilise : 
C = np.hstack((A, B)) #Concaténation horizontale de A et B
C.shape 

C = np.vstack((A, B)) #Concaténation verticale de A et B
C.shape

#On a egalement la possibilité de faire des concaténations avec des tableaux de dimensions différentes, mais il faut que les dimensions soient compatibles. 
#Par exemple, on peut concaténer un tableau de 2 lignes et 3 colonnes avec un tableau de 2 lignes et 2 colonnes, mais pas avec un tableau de 3 lignes et 2 colonnes.
np.concatenate((A, B), axis=0) #Concaténation verticale de A et B
#np.co      ncatenate((A, B), axis=1) #Concaténation horizontale de A et B, mais cela ne fonctionne pas car A et B n'ont pas le même nombre de lignes

## On a egalement la methode reshape qui permet de changer la forme d'un tableau.
D = np.array([[1, 2, 3], [4, 5, 6]]) #Tab de dim 2
D.shape
D.reshape(3, 2) #On change la forme de D en un tableau de 3 lignes et 2 colonnes
D.reshape(6, 1) #On change la forme de D en un tableau de 6 lignes et 1 colonne

### La methode ravel permet de transformer un tableau de n dimensions en un tableau de 1 dimension.
D.ravel() #On transforme D en un tableau de 1 dimension ou on l'applatit tout simplement 
```

## EXERCICE 
### Dans ce exercie on devra creer une fonction initialisation(m, n) avec m le nb lignes et n le nombre de colonnes. Elle va retourner une matrice aleatoire (m, n+1), avec une colonne biais (remplie de "1") tout a droite. 

```python
def initialisation(m, n) : 
    """
    Cette fonction permet d'initialiser un tableau de m lignes et n colonnes avec des valeurs aléatoires suivant une loi normale.
    """
    A = np.random.randn(m, n)
    A = np.concatenate((A, np.ones((A.shape[0], 1))), axis=1) #On ajoute une colonne de 1 à la fin du tableau A
    return A

print(initialisation(2, 3))
```
