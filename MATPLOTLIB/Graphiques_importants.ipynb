## Graphique 3D avec plt.scatter()

Pour faire ce genre de graphiques 3D qui viennent du sous-paquet **mpl_toolkits.mplot3d**, il faut d'abord charger les graphiques 

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
```

```python
#Ensuite on cree un objet qui va contenir notre figure 3D
ax = plt.axes(projection='3d')
ax.scatter(1, 2, 3, color='r', s=100)  # Exemple de point rouge dans l'espace 3D
```

### Visualisation d'une surface 3D et des fonctions de maths qu'on aurait modelise 

```python
f = lambda x, y: np.sin(x) + np.cos(x+y)
```

```python
X = np.linspace(0, 5, 100)
Y = np.linspace(0, 5, 100)
X, Y = np.meshgrid(X, Y) # Création d'une grille de points 2D avec meschgrid
Z = f(X, Y)
```

```python
Z.shape
```

```python
ax = plt.axes(projection='3d')
ax.plot_surface(X, Y, Z, cmap= 'plasma') #cmap = 'plasma' permet de choisir la palette de couleur
```

## Contour Plots

Il permet notammement de mieux comprendre les problemes d'optimisation 

```python
plt.contour(X, Y, Z, 20) ## le valeur 20 represente juse le niveau de precision qu'on aimerais donner a notre graphique 
```
