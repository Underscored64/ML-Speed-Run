**Il existe deux methodes en matplotlib pour creer de graphiques**

### La methode Classique et la methode avec POO

### la methode plot()

Pour construire un graphe de maniere simple avec matplotlib on utilise la fonction plot(x, y) avec x et y l'ensemble des coordonnees pour tracer le graphe 

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2, 10)
y = x ** 2
print(x)
```

```python
#x e t y doivent etre de meme taille pour pouvoir les tracer sur un graphique
plt.plot(x, y)
plt.show()
```

### La mtehode scatter() pour le tracer des nuages de points

```python
plt.scatter(x, y)
plt.show()
```

### Personnaliser son graphique 

Les possibilites sont infinies mais on ne va voir que quelques uns

- **c** pour la couleur : 'red', 'green', 'blue'
- **lw** pour l'epaisseur du trait
- **ls** style du trait : '--'

```python
plt.plot(x, y, c = 'red', lw = 3, ls = '--')
plt.show()
```

## Le cycle de vie d'une figure matplotlib

```python
'''Cycle de vie d'une figure matplotlib
1- Creation de la figure avec plt.figure() (la fenetre sur laquelle on va travailler)
2- Tracer des courbes avec plt.plot() ou plt.scatter()
3- Ajouter les legendes, les titres, les axes avec plt.title(), plt.xlabel(), plt.ylabel()...
4- Afficher la figure avec plt.show()
'''

plt.figure() #On peut itiliser l'attribut figsize pour definir la taille de la figure
plt.plot(x, y, label = 'quadratique') 
plt.plot(x, x**3, c = 'red', label = 'cubique')
plt.title('Figure 1')
plt.xlabel('axe x')
plt.ylabel('axe y')
plt.legend()
plt.show()
#Pour maontenant sauvegarder la figure on fait :
plt.savefig('figure1.png') #On peut definir le format de la figure avec l'extension du fichier
```

### On peut egalement definir une grille graphique sur notre figure (plusieurs graphiques)

```python
# La methode utiliser est subplot()
# plt.subplot(lignes, colonnes, position
plt.subplot(2, 1, 1) 
plt.plot(x, y, c = 'red')
plt.title('graphique 1')
plt.subplot(2, 1, 2)
plt.plot(x, y, c = 'blue')
plt.title('graphique 2')
```
